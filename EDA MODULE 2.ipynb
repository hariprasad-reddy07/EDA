```python
import pandas as pd

indA = pd.Index([1, 3, 5, 7, 9])
indB = pd.Index([2, 3, 5, 7, 11])

print("Index A:")
print(indA)

print("\nIndex B:")
print(indB)

intersection = indA.intersection(indB)

union = indA.union(indB)

symmetric_difference = indA.symmetric_difference(indB)

print("\nIntersection:")
print(intersection)

print("\nUnion:")
print(union)

print("\nSymmetric Difference:")
print(symmetric_difference)
```

```python
import pandas as pd
population = pd.Series(
    [124.8, 61.1, 104.1, 68.5, 91.1],
    index=["Maharashtra", "Karnataka", "Bihar", "Punjab", "Uttar Pradesh"]
)
area = pd.Series(
    [307713, 191791, 94163, 50362, 240928],
    index=["Maharashtra", "Karnataka", "Bihar", "Punjab", "Uttar Pradesh"]
)
df = pd.DataFrame({
    "Population": population,
    "Area": area
})
df["Population Density"] = df["Population"] / df["Area"]

print(df)
```

```python
df["Population Density"] = (df["Population"] * 100000) / df["Area"]

print(df)
```

```python
df["Population Density"] = (df["Population"] * 100000) / df["Area"]

print(df.round(2))
```
