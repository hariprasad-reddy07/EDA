```python
import numpy as np
import pandas as pd

data4 = pd.DataFrame([[1, np.nan, 2], [2, 3, 5], [np.nan, 4, 6]])
print(data4)


print(data4.fillna(0))
print(data4.ffill())
print(data4.bfill())
```

```python
import pandas as pd

index = [('A', 2010), ('A', 2020), ('A', 2025), ('B', 2010), ('B', 2020), ('B', 2025)]
population = [123, 456, 789, 101, 121, 131]

# Create the pandas Series first before reindexing
pop = pd.Series(population)

i = pd.MultiIndex.from_tuples(index)
print(i)

pop = pop.reindex(i)
print(pop)

pop_df = pop.unstack()
print(pop_df)
pop_df.stack()
```

```python
import pandas as pd


states = ['A', 'A', 'A', 'B', 'B', 'B']
years = [2010, 2020, 2025, 2010, 2020, 2025]
population = [123, 456, 789, 101, 121, 131]
i = pd.MultiIndex.from_arrays([states, years], names=['state', 'year'])
print(i)

pop = pd.Series(population, index=i)
print(pop)
pop_df = pop.unstack()
print(pop_df)
```

```python
import pandas as pd

states = ['A', 'B']
years = [2010, 2020, 2025]
population = [123, 456, 789, 101, 121, 131]

i = pd.MultiIndex.from_product([states, years], names=['state', 'year'])
print(i)

pop = pd.Series(population, index=i)
print(pop)
pop_df = pop.unstack()
print(pop_df)
```
