```python
import numpy as np
import pandas as pd
population=pd.Series({'California':78983890,'Texas':84368493,'New York':36536986,'Florida':97655373})
print(population)

df1=pd.DataFrame(population,columns=['population'])
print(df1)

df2=pd.DataFrame(np.random.rand(3,2),columns=['foo','bar'],index=['a','b','c'])
print(df2)

df3=pd.DataFrame([{'a':1,'b':2},{'b':3,'c':4}])
print(df3)

d=[
    {'a':i,'b':2*i}
    for i in range(10)
]
df4=pd.DataFrame(d)
print(df4)
```

```python
import pandas as pd
s1 = pd.Series([1, 2, ], index=['A', 'B'])
s2 = pd.Series([4, 5, 6], index=['A', 'B', 'C'])
s3 = pd.Series([7, 8, 9], index=['A', 'B', 'C'])
series_list = [s1, s2, s3]
df_rows = pd.DataFrame(series_list)
print(df_rows)
```

```python
import numpy as np
import pandas as pd
ind=pd.Index([2,3,5,7,11])
print(ind)
print(ind.size,ind.shape,ind.ndim,ind.dtype)
print(ind[1])
print(ind[::2])
print(ind[1:4:2])
indA=pd.Index([1,3,5,7,9])
indB=pd.Index([2,3,6,9,11])
print(indA & indB)
print(indA ^ indB)
print(indA | indB)
print(indA.insertsection(indB))
```

```python
import pandas as pd

population = pd.Series(
    [20411000, 32941000, 12765000, 11503000, 14974000],
    index=["Mumbai", "Delhi", "Bengaluru", "Chennai", "Kolkata"]
)

print("Population Series:")
print(population)

print("\nPopulation of Bengaluru:")
print(population["Bengaluru"])

print("\nDelhi to Chennai:")
print(population["Delhi":"chennai"])
```
